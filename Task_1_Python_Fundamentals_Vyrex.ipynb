```python
# TASK 1: FIND THE LARGEST NUMBER IN A LIST

def find_largest(numbers):

    largest = numbers[0]

    for num in numbers:

        if num > largest:
            largest = num

    return largest


numbers = [10, 25, 5, 40, 15]

result = find_largest(numbers)

print("Original List:", numbers)
print("Largest Number:", result)
```

```python
# Program 2: Check Even or Odd

def check_even_odd(number):

    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"


num = 15

result = check_even_odd(num)

print("Number:", num)
print("Result:", result)
```

```python
# Program 3: Calculate Factorial

def factorial(number):

    result = 1

    for i in range(1, number + 1):
        result = result * i

    return result


num = 5

print("Factorial:", factorial(num))
```

```python
# Program 4: Count Vowels

def count_vowels(text):

    vowels = "aeiou"
    count = 0

    for char in text.lower():

        if char in vowels:
            count += 1

    return count


sentence = "Machine Learning"

result = count_vowels(sentence)

print("Sentence:", sentence)
print("Total vowels:", result)
```

```python
# Program 5: Remove Duplicate Elements

def remove_duplicates(numbers):

    unique_numbers = []

    for num in numbers:

        if num not in unique_numbers:
            unique_numbers.append(num)

    return unique_numbers


numbers = [10, 20, 10, 30, 20, 40, 50]

result = remove_duplicates(numbers)

print("Original List:", numbers)
print("After Removing Duplicates:", result)
```
